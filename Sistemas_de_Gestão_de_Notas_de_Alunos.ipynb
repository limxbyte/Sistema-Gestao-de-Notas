```python

# Sistema de Gestão de Notas de Alunos

# Função para cadastrar as notas
def cadastrar_notas():
    notas = []

    # Pergunta quantas notas serão cadastradas
    quantidade = int(input("Quantas notas deseja cadastrar? "))

    # Repete a pergunta para cada nota
    for i in range(quantidade):
        nota = float(input(f"Digite a nota {i + 1}: "))
        notas.append(nota)

    return notas


# Função para calcular a média
def calcular_media(notas):
    media = sum(notas) / len(notas)
    return media


# Função para verificar a situação do aluno
def verificar_situacao(media):
    if media >= 7:
        return "Aprovado"
    else:
        return "Reprovado"


# Função para exibir o relatório final
def exibir_relatorio(notas, media, situacao):
    print("\n===== RELATÓRIO FINAL =====")
    print("Notas:", notas)
    print(f"Média: {media:.2f}")
    print("Situação:", situacao)


# Programa principal
notas = cadastrar_notas()
media = calcular_media(notas)
situacao = verificar_situacao(media)

exibir_relatorio(notas, media, situacao)
```
